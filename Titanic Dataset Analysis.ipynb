### Part B: Filtering Data

- Display all passengers whose age is greater than 50.

```python
print(df[df['age'] > 50])
```

- Display all female passengers.

```python
print(df[df['gender'] == 'female'])
```

- Display passengers who:
    * Survived

```python
print(df.query('survived == 1'))
```

 * Were in First Class (`Pclass = 1`)

```python
print(df.query('pclass == 1'))
```

- Display passengers whose fare was greater than 100.

```python
print(df.query('fare > 100').head())
```

- Using `query()`, display passengers with:

   * Age > 30
   * Fare > 50

```python
print(df.query('age > 30'))
```

```python
print(df.query('fare > 50'))
```

- Display all passengers embarking from Southampton (`Embarked = 'S'`).

```python
print(df[df['embarked'] == 'S'])
```

### Part E: Missing Values

- Count missing values in each column.

```python
print(df.isnull().sum())
```

- Find how many missing ages exist.

```python
print(df['age'].isnull().sum())
```

- Create a new DataFrame where missing ages are replaced with the mean age.

```python
df_new = df.copy()
df_new['age'] = df['age'].fillna(df['age'].mean())
#print(df_new.isnull().sum())
print(df_new)
```

- Create another DataFrame after removing all rows containing missing values.

```python
df_clean = df.dropna()
#print(df_clean.isnull().sum())
print(df_clean)
```

### Part F: Grouping

- Calculate the average age of passengers in each class.

```python
print(df.groupby('pclass')['age'].mean())
```

- Calculate the average fare paid in each class.

```python
print(df.groupby('pclass')['fare'].mean())
```

- Find the survival rate by gender.

```python
print(df.groupby('gender')['survived'].mean())
```

- Find the average age of males and females.

```python
print(df.groupby('gender')['age'].mean())
```

- Find the number of passengers embarking from each port.

```python
print(df.groupby('embarked')['embarked'].count())
```
